```python
from __future__ import print_function
import numpy as np
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers.core import Dense, Dropout, Activation
from keras.optimizers import SGD
from keras.utils import np_utils
np.random.seed(1671) # network and training 
NB_EPOCH = 10 
BATCH_SIZE = 128
VERBOSE = 1
NB_CLASSES = 10 #number of outputs = number digits
OPTIMIZER = SGD() # optimizer, explained later in this chapter
N_HIDDEN = 128 

VALIDATION_SPLIT=0.2 #how much TRAIN is reserved for VALIDATION
# data: shuffles and split between train and test sets
(X_train, y_train), (X_test, y_test) = mnist.load_data()
#X_train is 60000 rows of 28x28 values --> reshape in 60000 x 784
RESHAPED = 784
#
X_train = X_train.reshape(60000, RESHAPED)
X_test = X_test.reshape(10000, RESHAPED)
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
# normalize
X_train /= 255
X_test /= 255
print(X_train.shape[0], 'train samples')
print(X_test.shape[0], 'test samples')
# convert class vectors to binary class matrices
Y_train = np_utils.to_categorical(y_train, NB_CLASSES)
Y_test = np_utils.to_categorical(y_test, NB_CLASSES)
# M_HIDDEN hidden layers
# 10 outputs
# final stage is softmax
model = Sequential()
model.add(Dense(N_HIDDEN, input_shape=(RESHAPED,)))
model.add(Activation('relu'))
model.add(Dense(N_HIDDEN))
model.add(Activation('relu'))
model.add(Dense(NB_CLASSES))
model.add(Activation('softmax'))
model.summary()
model.compile(loss='categorical_crossentropy',
              optimizer=OPTIMIZER, 
              metrics=['accuracy']) 
history = model.fit(X_train, Y_train,
                    batch_size=BATCH_SIZE,
                    epochs=NB_EPOCH,
                    verbose=VERBOSE,
                    validation_split=VALIDATION_SPLIT)
score = model.evaluate(X_test, Y_test, verbose=VERBOSE)
print("Test score:", score[0])
print('Test accuracy:', score[1])
```

As you can see, the gradual increase in accuracy is noticable between 10, 20 and 200 epochs

We start with a baseline of 93% which could use improvement. The second test we ran double the epochs and achieved an accuracy increase of about 1.2%. Finally, with 200 epochs we had an increase of about 3.5% which is about 4.7% from baseline.

In our first test, the fact that the training accuracy is lower than the test accuracy tells use the the network is not getting enough time to train. The second test, we can see that the training accuracy is only slightly better that the test accuracy, this tells us that we should try giving it some more time. and finally, in the third test we have a substantial increase, this test is much more accurate than the other two.

The downside to using 200 epochs is the amount of time lost while training. Training 200 epochs runs about 10 fold of 20 epochs.

There are some ways to get the number closer to 100% for instance switching the optimizer to adam and setting a 30% dropout would help. we could also increase the hidden nuerons, this would significantly increase our training time as well.
